## Research Question

Do neighborhoods surrounding professional sports stadiums in the US have a different economic makeup than the rest of their host city? Specifically, is there a significant difference in median household income (wealth), employment rates (opportunity), and property values (wealth) between neighborhoods within a 2-mile radius of a stadium and those further away within the same city?

## Background and Prior Work

Professional sports stadiums are often significantly large venues that host often profitable sporting events for the NFL, NBA, NHL, and other sporting leagues. Stadiums are also culturally significant as events like the Super Bowl, hosted in the Caesars Superdome, New Orleans, Louisiana in 2025, had around 120 million viewers and around 76,000 filled seats.It can be assumed that stadiums  stimulate economic activity around the local area as businesses take advantage of the large foot traffic stadiums create, creating more jobs, increasing spending, tourism, and property value <sup>[2](#https://www.researchgate.net/publication/46552225_How_Does_a_New_Sports_Stadium_Affect_Housing_Values_The_Case_of_Fedex_Field)</sup>. 


However, studies and articles conclude that the construction of new stadiums yields negligible improvements to local employment<sup>[3](#https://doi.org/10.1002/pam.22534)</sup>, and may take spending away from local businesses as people spend it on stadium related costs instead. Additionally, stadium construction is often funded by public subsidies that can be used on other local construction projects or programs. While stadiums may generate a localized economic surge within their immediate vicinity<sup>[1](#https://www.aimspress.com/article/doi/10.3934/NAR.2024024?viewType=HTML)</sup> and foster civic identity and pride, analysts argue their regional economic impact is limited.


For this project, we are observing how professional sports stadiums affect the local economy. In a 2024 research paper, Antolini Valerio analyzes trends in GDP per capita in Turin, Italy around the construction of a new and technologically impressive NFL stadium called a next-generation stadium. The writer found a causal relationship between an increase in GDP per capita with the new stadium compared to other regions of Italy that showed comparatively lower growths in GDP per capita1. The paper provides some insights on how new and expensive stadiums can change local economic production. In a study by Charles Tu, the construction of a stadium causes homes to be sold at a lower price, but over longer periods of time land value increases. The study shows that there are changes to local property values which will be examined and compared in our own data. The paper by Bradbury and co, details the large public costs of newly constructed professional sports stadiums compared to how little economic improvement they provide. This review on the stadiums and their impacts show the confounding ethical, public, and economic variables that we need to interpret to understand how professional sports stadiums impact localized economies.

## Hypothesis


We predict that neighborhoods within a 2 mile radius of a professional sports stadium will have a significantly different economic profile than the rest of their host city, characterized by higher commercial activity but lower residential wealth. Specifically, we predict that stadium adjacent neighborhoods will have higher employment rates due to the localized concentration of hospitality, retail, and stadium operations. However, we also predict that median property values and median household incomes will be lower in this immediate 2 mile radius compared to the city wide baseline. We attribute this to the negative residential externalities of large venues, like traffic congestion, noise, and overcrowding, which make the immediate area less desirable for high income homebuyers, pushing residential wealth into quieter neighborhoods further away.

2. Major concerns: This dataset is from 2021 and may not reflect accurate team locations due to the movement of sports teams within the last 5 years. Additionally, its exhaustiveness (i.e., whether or not it includes all professional sports stadiums) has not been confirmed.

```python
import pandas as pd
import numpy as np
import os

# Setup
raw_data_path = 'data/00-raw/stadiums.csv'
processed_data_path = 'data/02-processed/processed_data.csv'

os.makedirs('data/02-processed', exist_ok=True)

try:
    df_raw_stadiums = pd.read_csv(raw_data_path, low_memory=False)
except FileNotFoundError:
    print('Files not found. Please ensure they are in data/00-raw/')
    df_raw_stadiums = pd.DataFrame()

# Tidy Structure
print('First 5 Rows:')
print(df_raw_stadiums.head())
print('Column Names:')
print(df_raw_stadiums.columns.tolist())
print('Each row represents one professional sports team and each column represents a variable, such as league and coordinates.')
print('Dataset is already tidy.')

# Shape
print('Dataset Shape (rows, columns):', df_raw_stadiums.shape)

# Missing Values
missing_vals = df_raw_stadiums.isna().sum()
print('Missing Values Per Column:')
print(missing_vals)
print('Missingness Correlation Matrix:')
print(df_raw_stadiums.isna().astype(int).corr())

# Location Outliers
numeric_cols = ['Lat', 'Long']
for col in numeric_cols:
    Q1 = df_raw_stadiums[col].quantile(0.25)
    Q3 = df_raw_stadiums[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df_raw_stadiums[(df_raw_stadiums[col] < lower) | (df_raw_stadiums[col] > upper)]
    print(f'Outliers in {col}: {len(outliers)}')
# outliers in {col}: {len(outliers)}')

# Data Cleaning
df_clean_stadiums = df_raw_stadiums.dropna(subset=['Lat', 'Long'])

# Remove Canadian teams (falling outside US Census ZCTA boundaries)
canadian_teams = [
    'Toronto Raptors', 'Montreal Canadiens', 'Winnipeg Jets', 
    'Ottawa Senators', 'Vancouver Canucks', 'Edmonton Oilers', 
    'Toronto Maple Leafs', 'Calgary Flames', 'Toronto Blue Jays', 
    'Vancouver Whitecaps FC', 'Toronto FC', 'CF Montréal'
]
df_clean_stadiums = df_clean_stadiums[~df_clean_stadiums['Team'].isin(canadian_teams)]

print('Final Dataset Shape After Cleaning:', df_clean_stadiums.shape)

# Save Processed Data
df_clean_stadiums.to_csv(processed_data_path, index=False)
print('Processed dataset saved to data/02-processed/')
```

### Mapping Stadiums to Cities and ZIP Codes
Connects raw stadium coordinates to specific geographic identifiers: Zip Codes and City Names

- Spatial Alignment: Convert latitude and longitude coordinates into point objects and perform a spatial join with Census ZCTA boundaries to identify which ZIP code each stadium is in

- Data Enrichment: Merge the spatial results with data from uszips to attach official city and state names

```python
#load uszips
df_uszips = pd.read_csv('data/00-raw/uszips.csv', compression='zip')

df_uszips['zip'] = df_uszips['zip'].astype(str).str.zfill(5)

df_uszips = df_uszips[['zip', 'city', 'state_id']]
```

```python
import geopandas as gpd
import pandas as pd

# vectorized geometry creation
gdf_stadiums = gpd.GeoDataFrame(
    df_clean_stadiums, 
    geometry=gpd.points_from_xy(df_clean_stadiums.Long, df_clean_stadiums.Lat),
    crs="EPSG:4326" 
)

# Load the shapes
gdf_zcta_shapes = gpd.read_file('zip://data/00-raw/tl_2025_us_zcta520.zip')

# Project the points
gdf_stadiums = gdf_stadiums.to_crs(gdf_zcta_shapes.crs)

# Spatial Join
stadiums_with_zips = gpd.sjoin(gdf_stadiums, gdf_zcta_shapes, how="left", predicate="within")

# Format ZIPs for merging
stadiums_with_zips['ZCTA5CE20'] = stadiums_with_zips['ZCTA5CE20'].astype(str)
df_uszips['zip'] = df_uszips['zip'].astype(str).str.zfill(5)

# Left Join to Kaggle reference
final_stadiums = stadiums_with_zips.merge(
    df_uszips[['zip', 'city', 'state_id']], 
    left_on='ZCTA5CE20', 
    right_on='zip', 
    how='left'
)

# Clean up and Rename
final_stadiums = final_stadiums.drop(columns=['zip'])
final_stadiums = final_stadiums.rename(columns={
    'ZCTA5CE20': 'Zip_Code', 
    'city': 'City', 
    'state_id': 'State'
})

# Apply critical patches
final_stadiums.loc[final_stadiums['Team'] == 'Vegas Golden Knights', ['City', 'State']] = ['Las Vegas', 'NV']
final_stadiums.loc[final_stadiums['Team'] == 'FC Dallas', ['City', 'State']] = ['Frisco', 'TX']

final_stadiums['City_State_Unique'] = final_stadiums['City'] + ", " + final_stadiums['State']

# Drop unnecessary columns
final_stadiums = final_stadiums.drop(columns=[
    'index_right', 'GEOID20', 'GEOIDFQ20', 'CLASSFP20', 
    'MTFCC20', 'FUNCSTAT20', 'INTPTLAT20', 'INTPTLON20'
])

print(f"Number of Stadiums: {len(final_stadiums)}")

missing_zips = final_stadiums['Zip_Code'].isna().sum()
missing_cities = final_stadiums['City'].isna().sum()
print(f"\nMissing ZIP Codes: {missing_zips}")
print(f"Missing Cities: {missing_cities}")

display(final_stadiums.head(10))
```

### Defining the Treatment Group:
Isolate the specific neighborhoods that are most likely to be affected by the presence of a professional sports stadium

- Spatial Buffer Analysis: Create a standardized 2 mile zone around each stadium coordinate to define the boundary for potential localized economic effects

- Neighborhood Identification: Determine which ZIP codes fall within these impact zones

- Final Treatment Group: Create list of Treatment ZIP codes

```python
import geopandas as gpd

# Project to meters and create 2 mile buffer
treatment_zones = final_stadiums.to_crs(epsg=5070)
treatment_zones['geometry'] = treatment_zones.geometry.buffer(3218.68)

# Project zones back to match the Census ZCTA coordinate system
treatment_zones_aligned = treatment_zones.to_crs(gdf_zcta_shapes.crs)

# Clean columns
cols_to_drop = ['index_right', 'index_left']
treatment_zones_aligned = treatment_zones_aligned.drop(columns=[c for c in cols_to_drop if c in treatment_zones_aligned.columns])

# Find all Census ZCTAs that touch the 2 mile stadium boundaries
stadium_neighborhoods = gpd.sjoin(
    gdf_zcta_shapes, 
    treatment_zones_aligned, 
    how="inner", 
    predicate="intersects"
)

# Extract unique ZIPs
unique_treatment_zips = stadium_neighborhoods['ZCTA5CE20'].unique()
treatment_group = pd.DataFrame({'Treatment_Zip': unique_treatment_zips})

# Ensure all ZIPs are 5 char strings
treatment_group['Treatment_Zip'] = treatment_group['Treatment_Zip'].astype(str).str.zfill(5)

print(f"Treatment Group defined with {len(treatment_group)} unique ZIP codes.")
```

### Defining the Control Group
Identify a baseline for comparison by finding neighborhoods within the same urban economies that are far enough away to be unaffected by stadium externalities.

- Identify Stadium Cities and Associated Zips: Create a master list of the 71 unique cities (and all the Zips they contain) that host the professional stadiums in the study

- Exclude the Treatment Group: Remove any neighborhoods that fall within the 2 mile stadium radius

- Final Control Group: Create List

```python
# 1. Identify all ZIPs belonging to your 71 stadium cities
stadium_cities = final_stadiums['City_State_Unique'].unique()
df_uszips['City_State_Unique'] = df_uszips['city'] + ", " + df_uszips['state_id']
city_zips = df_uszips[df_uszips['City_State_Unique'].isin(stadium_cities)].copy()
city_zips['zip'] = city_zips['zip'].astype(str).str.zfill(5)

# 2. Extract the treatment list for filtering
treatment_zip_list = treatment_group['Treatment_Zip'].unique()

# 3. Create control group by excluding treatment ZIPs from the city list
control_group = city_zips[~city_zips['zip'].isin(treatment_zip_list)].copy()

# 4. Clean and Rename
control_group = control_group.rename(columns={'zip': 'Control_Zip'})

print("Control Group Summary:")
print(f"Total Stadium Cities: {len(stadium_cities)}")
print(f"Total ZIPs in these cities: {len(city_zips)}")
print(f"Treatment ZIPs within these cities: {len(city_zips[city_zips['zip'].isin(treatment_zip_list)])}")
print(f"Control ZIPs: {len(control_group)}")
```

```python
# Label the groups for the final analysis
treatment_group['Group'] = 'Treatment (2-Mile Radius)'
control_group['Group'] = 'Control (Rest of City)'

# Standardize column names for stacking
t_list = treatment_group.rename(columns={'Treatment_Zip': 'zip_code'})[['zip_code', 'Group']]
c_list = control_group.rename(columns={'Control_Zip': 'zip_code'})[['zip_code', 'Group']]

# Stack them into one master lookup table
combined_groups = pd.concat([t_list, c_list], ignore_index=True)

# Final safety check: ensure no duplicates if a ZIP was somehow in both lists
combined_groups = combined_groups.drop_duplicates(subset=['zip_code'])

print(f"Master Key 'combined_groups' created with {len(combined_groups)} total ZIPs.")
print(combined_groups['Group'].value_counts())
```

### Dataset #2: ACS Demographic and Housing Estimates
The most important metrics in the ACS Demographic and Housing Estimates dataset include total population, race and ethnicity composition, and total housing units. Total population is measured as an estimated count of individuals living in a given geographic area (such as a census tract, city, or state). For example, a census tract might have an estimated population of 4,200 people. These counts allow us to compare neighborhood size and calculate per-capita measures when analyzing economic outcomes.

Race and ethnicity variables are also reported as counts and percentages, showing the share of residents identifying as White, Black or African American, Asian, Hispanic or Latino, and other categories. Changes in these percentages over time can signal demographic shifts such as migration, redevelopment, or possible displacement.

Total housing units is measued as a count of housing units in a specified area.

There are several major concerns with this dataset. First, because the ACS is survey based, estimates are subject to sampling error and potential nonresponse bias. Certain populations, such as undocumented residents or lower-income households may be undercounted. Second, smaller geographic units like census tracts often have larger margins of error, which can reduce confidence in precise comparisons. Finally, the dataset provides aggregated summary statistics rather than individual level data, meaning we cannot directly observe displacement or track specific residents over time.

Additionaly, demographic changes observed near stadiums may reflect broader citywide economic trends rather than the direct impact of the stadium itself, making careful statistical design necessary to avoid misleading conclusions that may be caused by other factors.

Note: The loading and parsing shown here is done on all ZIP Code tabulation areas in the united states.

```python
import pandas as pd
from pathlib import Path

raw_data_path = "data/00-raw/ACSDP5Y2024.DP05-Data.csv"
metadata_path = "data/00-raw/ACSDP5Y2024.DP05-Column-Metadata.csv"

df_raw = pd.read_csv(raw_data_path, encoding="utf-8-sig", low_memory=False)

cols_to_keep = {
    "GEO_ID": "geo_id",
    "NAME": "area_name",
    "DP05_0001E": "total_population",
    "DP05_0037E": "white_alone",
    "DP05_0045E": "black_alone",
    "DP05_0061E": "asian_alone",
    "DP05_0090E": "hispanic_or_latino",
    "DP05_0105E": "total_housing_units",
    "DP05_0105M": "moe_total_housing_units"
}

df = df_raw[list(cols_to_keep.keys())].rename(columns=cols_to_keep)
df = df.iloc[1:].reset_index(drop=True)
df["zip_code"] = df["area_name"].str.extract(r"(\d{5})")

numeric_cols = ["total_population","white_alone","black_alone","asian_alone","hispanic_or_latino","total_housing_units","moe_total_housing_units"]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=["total_population"])
df["pct_white"] = df["white_alone"] / df["total_population"]
df["pct_black"] = df["black_alone"] / df["total_population"]
df["pct_asian"] = df["asian_alone"] / df["total_population"]
df["pct_hispanic"] = df["hispanic_or_latino"] / df["total_population"]

df_clean = df[[
    "zip_code",
    "total_population",
    "white_alone",
    "black_alone",
    "asian_alone",
    "hispanic_or_latino",
    "pct_white",
    "pct_black",
    "pct_asian",
    "pct_hispanic",
    "total_housing_units",
    "moe_total_housing_units"
]]

Path("data/01-clean").mkdir(parents=True, exist_ok=True)
df_clean.to_csv("data/01-clean/acs_zip_demographics_clean.csv", index=False)

print(df_clean.shape)
```

### Dataset #3: Selected Economic Characteristics of County Subdivisions

# 1. Industry

This variable describes the sector in which employed individuals work. Examples include:

1. Agriculture, forestry, fishing, and mining
2. Construction
3. Manufacturing
4. Retail trade
5. Professional, scientific, and management services
6. Education and healthcare services
7. Public administration

Industry is reported as a count of workers or as a percentage of the employed population in each sector.

This metric helps describe the economic structure of a subdivision. For example:
- A high percentage in agriculture may indicate a rural economy.
- A high percentage in professional services may indicate an urban or higher-income area.

# 2. Employment Status

This variable categorizes the working-age population into groups such as:
- Employed
- Unemployed (actively seeking work)
- Not in labor force

Values are reported as counts or percentages (%) of the civilian labor force.

The unemployment rate is particularly important:
- In stable economic conditions, unemployment is often around 3–6%.
- Rates significantly above 8–10% may indicate local economic distress.

Labor force participation rate is also included, which measures the percentage of the population that is either working or actively seeking work.

# 3. Commuting to Work

This category describes how people travel to work and includes:
- Car, truck, or van
- Public transportation
- Walked
- Worked from home
- Other means

It also includes average commute time (minutes).

Units are:
- Percent of workers by transportation mode (%)
- Mean commute time in minutes
- This variable gives insight into infrastructure, urbanization, and economic integration with nearby regions.

# 4.Occupation

Occupation refers to the type of job performed, such as:
- Management, business, science, and arts occupations
- Service occupations
- Sales and office occupations
- Natural resources, construction, and maintenance occupations
- Production, transportation, and material moving occupations

Values are usually percentages of employed persons.

Occupation differs from industry:
1. Industry describes what sector the employer is in.
2. Occupation describes what type of job the person performs.

For example, an accountant working for a hospital is:
- Industry: Healthcare

- Median household income (USD per year)
- Per capita income (USD per year)
- Mean earnings (USD per year)
- Receipt of public assistance (%)
- Social Security income (%)

Income values are measured in U.S. dollars per year, and median household income is particularly important:
- National median income in recent years is around $70,000–$75,000.

- Values above $100,000 often reflect affluent areas.

6. Health Insurance Coverage

This variable measures:
- Percent insured
- Percent uninsured

Units are percentages (%).

The uninsured rate is important:
- National uninsured rates are often around 8–10%.
- Rates above 15–20% may indicate limited access to healthcare.

This metric is closely tied to employment patterns, income, and state-level Medicaid expansion policies.

7. Percentage of Families and People Below the Poverty Level

This variable measures the percentage of:
- Individuals below poverty
- Families below poverty
- Units: percentage (%)

The federal poverty threshold is defined annually by the federal government and depends on family size and composition.

Interpretation:
- U.S. poverty rate is typically around 10–15%.
- Rates above 20% may indicate significant economic vulnerability.
- Extremely low poverty rates (<5%) often occur in affluent suburbs.

This metric is one of the most important indicators of economic distress.

# Concerns with the Dataset

## 1. Sampling Error (ACS-Based Estimates)
If this dataset comes from the American Community Survey (ACS), it is based on a sample rather than a full census. Smaller subdivisions often have:
- Larger margins of error
- Less stable estimates
- Small population areas may show exaggerated variation.

## 2. Suppression or Missing Data
Some Zip Codes may have missing values due to:
- Small population size
- Confidentiality protection
- Insufficient survey responses

## 3. Self-Reported Data
Income, employment status, and insurance coverage are self-reported, which introduces potential reporting bias.

## 4. Comparability Across Zip Codes

Zip Codes do not determine the number of people in that area, and some zip codes can be inflated and have many more people, or a skewed wealth distribution.

### Data Set #4: ACS 2024 5-Year Estimates Selected Housing Characteristics (DP04)

This dataset is from the U.S. Census Bureau’s American Community Survey (ACS) 2024 5-Year Estimates. It provides a comprehensive profile of the housing stock, financial costs of homeownership and renting, and physical characteristics of housing units across specific geographies. The ACS is an ongoing survey that provides characteristics data, helping planners, researchers, and real estate analysts understand the quality of life and economic burden of housing in specific regions.

The dataset includes several critical metrics for housing market analysis. Median Home Value and Median Gross Rent are provided in US Dollars, they represent the midpoint of all reported values, where half are higher and half are lower. A healthy housing market typically sees these values rising in line with local inflation, whereas stagnant or falling values may indicate economic decline. Housing Tenure (Owner vs. Renter occupied) is measured in raw counts and percentages, a high owner-occupancy rate suggests neighborhood stability, while high renter-occupancy can indicate a transient population or an area with high density multi-family development. Another important metric is Gross Rent as a Percentage of Household Income (GRAPI). The Census Bureau generally considers households spending 30% or more of their income on housing to be "rent burdened," while those spending 50% or more are "severely rent burdened," indicating a potential crisis in affordability.

A major concern with this dataset is its reliance on self-reporting and sampling error. Because this is a survey of a sample of the population rather than a full count, every estimate comes with a Margin of Error (MOE). In smaller geographic areas (like specific census tracts near a stadium), the MOE can be quite large relative to the estimate, making the data less reliable. There is also a risk of Non-response Bias, when certain demographics, such as very high-income individuals or undocumented residents, may be less likely to respond to the survey accurately. Finally, "Median Value" is based on the respondent’s perception of what their home is worth, which may differ from actual market sale prices or tax appraisals.


```python
import pandas as pd
import numpy as np
import os

raw_data_path = 'data/00-raw/ACSDP5Y2024.DP04-Data.csv'

os.makedirs('data/02-processed', exist_ok=True)

df_raw_house = pd.read_csv(raw_data_path, low_memory=False)
    

cols_to_keep = {
    # Identifiers
    'GEO_ID': 'geo_id',
    'NAME': 'neighborhood_name',
    
    # Inventory
    'DP04_0001E': 'total_housing_units',
    'DP04_0002E': 'occupied_units',
    'DP04_0003E': 'vacant_units',
    
    # Building Age (New Construction Proxy)
    'DP04_0017E': 'built_2020_or_later',
    'DP04_0018E': 'built_2010_to_2019',
    'DP04_0019E': 'built_2000_to_2009',
    
    # Financials
    'DP04_0089E': 'median_home_value',
    'DP04_0134E': 'median_gross_rent',
    
    # Structure Type (Density Proxy)
    'DP04_0007E': 'single_family_detached',
    'DP04_0013E': 'large_apartments_20plus_units'
}


if 'DP04_0001E' in df_raw_house.columns:
    available_cols = [c for c in cols_to_keep.keys() if c in df_raw_house.columns]
    df_clean_house = df_raw_house[available_cols].rename(columns=cols_to_keep)
    
elif 'Label (Grouping)' in df_raw_house.columns:
    print("Detected Profile Format. Transposing...")
    df_clean_house = df_raw_house.set_index('Label (Grouping)').T
    df_clean_house.index.name = 'neighborhood_name'
    df_clean_house = df_clean_house.reset_index()
    
    pass 
else:
    df_clean_house = df_raw_house.copy()

def clean_census_number(val):
    if pd.isna(val): return np.nan
    s = str(val).strip()
    if s in ['(X)', '-', 'N', '**', '***', '(V)']: return np.nan
    s = s.replace(',', '').replace('+', '').replace('$', '').replace('±', '')
    try:
        return float(s)
    except:
        return np.nan

cols_to_convert = [c for c in df_clean_house.columns if c not in ['geo_id', 'neighborhood_name']]

for col in cols_to_convert:
    df_clean_house[col] = df_clean_house[col].apply(clean_census_number)

df_clean_house = df_clean_house.dropna(how='all')

if 'median_home_value' in df_clean_house.columns:
    print(f"Dropping {df_clean_house['median_home_value'].isna().sum()} rows with missing home values.")
    df_clean_house = df_clean_house.dropna(subset=['median_home_value'])

df_clean_house['zip_code'] = df_clean_house['neighborhood_name'].str.extract(r'(\d{5})')

output_path = 'data/02-processed/census_housing_char_clean.csv'
df_clean_house.to_csv(output_path, index=False)

print(f"Cleaned data saved to {output_path}")
print(f"Final Data Shape: {df_clean_house.shape}")
print(df_clean_house.head())
print(df_clean_house[['zip_code', 'median_home_value']].head())
```

```python
import scipy.stats as stats
import pandas as pd

stat_results = []

treatment_data = housing_analysis_df[housing_analysis_df['Group'] == 'Treatment (2-Mile Radius)']
control_data = housing_analysis_df[housing_analysis_df['Group'] == 'Control (Rest of City)']

for col in important_cols:
    # drop empty rows 
    t_values = treatment_data[col].dropna()
    c_values = control_data[col].dropna()
    
    # run Welch's T-Test 
    t_stat, p_val = stats.ttest_ind(t_values, c_values, equal_var=False)
    
    # check if the result is significant using 95% confidence threshold (p < 0.05)
    is_significant = "Yes" if p_val < 0.05 else "No"
    
    # save the results
    stat_results.append({
        'Metric': col,
        'T-Statistic': t_stat,
        'P-Value': p_val,
        'Significant (p < 0.05)': is_significant
    })

# dataframe
stats_df = pd.DataFrame(stat_results)
formatted_stats = stats_df.style.format({
    'T-Statistic': '{:.2f}',
    'P-Value': '{:.5f}'
})

display(formatted_stats)
```

### Visualizing the Economic "Stadium Effect"
These two charts show how professional sports stadiums alter the economic makeup and real estate landscape of their surrounding neighborhoods.

1. The Wealth Shift: Property Value Distribution (Left Chart)
- This density plot shows the concentration of neighborhood property values.

- The Rightward Shift: The gray curve (Treatment/Stadium neighborhoods) is shifted to the right compared to the green curve (Control/Rest of City). This represents the $105,700 premium we calculated earlier, proving that the baseline cost of living near a stadium is fundamentally higher.

- The Tail: The gray curve stretches much further out to the far right. This shows a higher concentration of expensive neighborhoods directly surrounding stadiums compared to the rest of the city.

2. Economic Makeup: Investment & Gentrification (Right Chart)
- This bar chart breaks down what percentage of a neighborhood's total housing supply is tied up in new development or sits empty, answering the economic makeup portion of our research question.

- High Vacancy (Investor Market): The gray bar on the left shows that stadium neighborhoods have a higher percentage of vacant housing units (>10%) than the rest of the city (~8%). This points to an investor driven market heavily populated by short term rentals (like game day Airbnbs) and secondary luxury homes rather than full time residents.

- Development Boom: The gray bar on the right shows a higher percentage of brand new construction (built 2020 or later) near stadiums, validating that these 2-mile radii are active hotspots for developers (often leading to gentrification).

```python
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.ticker as ticker

housing_analysis_df['vacancy_rate'] = (housing_analysis_df['vacant_units'] / housing_analysis_df['total_housing_units']) * 100
housing_analysis_df['new_build_rate'] = (housing_analysis_df['built_2020_or_later'] / housing_analysis_df['total_housing_units']) * 100

sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Plot 1: Wealth Distribution (KDE Plot)
sns.kdeplot(
    data=housing_analysis_df, 
    x='median_home_value', 
    hue='Group', 
    fill=True, 
    common_norm=False, 
    palette=['#95a5a6', '#2ecc71'], 
    alpha=0.5,
    ax=axes[0]
)

axes[0].set_title('Property Value Distribution\nDo Stadiums Shift the Whole Market?', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Median Home Value ($)', fontsize=12)
axes[0].set_ylabel('Density (Concentration of Neighborhoods)', fontsize=12)
axes[0].xaxis.set_major_formatter(ticker.StrMethodFormatter('${x:,.0f}'))

# Plot 2: Economic Makeup (Vacancy & New Builds)
rates_df = housing_analysis_df[['Group', 'vacancy_rate', 'new_build_rate']].melt(id_vars='Group', var_name='Metric', value_name='Percentage')

rates_df['Metric'] = rates_df['Metric'].replace({'vacancy_rate': 'Vacancy Rate', 'new_build_rate': 'New Build Rate (2020+)'})

sns.barplot(
    data=rates_df, 
    x='Metric', 
    y='Percentage', 
    hue='Group', 
    errorbar=None,
    palette=['#95a5a6', '#e74c3c'], 
    ax=axes[1]
)

axes[1].set_title('Neighborhood Makeup:\nInvestment Rates', fontsize=14, fontweight='bold')
axes[1].set_xlabel('')
axes[1].set_ylabel('Percentage of Total Housing Units (%)', fontsize=12)
axes[1].yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.1f}%'))

plt.tight_layout()
plt.show()
```

- High-End Market: The Treatment group maintains a visible "tail" extending toward the $2,000,000 mark, suggesting that stadiums are often located in or near luxury urban centers.

# Neighborhood Makeup

- Vacancy Rate: There is a significant difference here. Neighborhoods near stadiums have a vacancy rate of over 10%, whereas the rest of the city sits at approximately 8%. This supports our previous statistical analysis ($t = 4.14, p < 0.001$), suggesting these areas may have more short-term rentals (like Airbnbs) or speculative investments.

- New Build Rate (2020+): The Treatment group shows a slightly higher rate of new construction since 2020 (approx. 2.0%) compared to the Control group (approx. 1.8%). While the visual difference is small, our T-test ($t = 2.13, p = 0.033$) confirmed this is a statistically significant trend, indicating stadiums are hotspots for recent development.
