## Discusison and Conclusion

Our research found that neighborhoods within two miles of U.S. sports stadiums fundamentally contradict our initial hypotheses of job growth and lower property values. Instead, stadium proximity is linked to a roughly $105,700 premium on median home values (p<0.00001), significantly higher rents (p=0.00005), and increased new development and vacancy rates (p<0.05). Most notably, rather than boosting the local economy for residents, these stadium-adjacent neighborhoods face significantly higher average unemployment (6.79% vs. 5.84%, p=0.00002) and higher mean poverty rates (18.02% vs. 14.27%), while median household incomes remain statistically unchanged compared to the rest of the city (p=0.439).

Ultimately, these metrics point to the results of gentrification and speculative investment rather than broad community uplift. Stadium neighborhoods are more expensive to live in, attracting new development and short-term or vacant investor properties. However, this localized influx of wealth does not translate into economic prosperity for residents. Flat incomes combined with expensive housing and a 10% vacancy rate show a clear disconnect: the neighborhood is getting richer, but the residents are not. The people purchasing these expensive, newly developed properties likely do not live there full-time, leaving the long-term, local residents to navigate higher unemployment and poverty rates than the citywide average.

The broader skepticism found in our background literature aligned closely with these findings. Bradbury, Coates, and Humphreys (2024) argued that the economic improvements promised by new stadiums are often vastly overstated compared to their massive public costs, a sentiment our unemployment data strongly supports. Furthermore, our property valuation findings add a modern layer to Charles Tu’s (2005) observations on stadium-adjacent real estate. While Tu noted that homes might initially sell for less before land values eventually rise, our 2024 ACS data suggests we are currently observing the end of that timeline. Stadiums appear to act as anchors for luxury urban development and out-of-town investment, echoing Valerio's (2024) insights on stadiums driving localized economic production in a way that prioritizes real estate capital over resident welfare.

This highlights an ethical issue regarding urban planning and public finance. Professional sports stadiums are frequently subsidized by taxpayer dollars under the promise of job creation and community revitalization. However, our data suggests a failure to deliver on this promise for the immediate community. The areas directly surrounding these publicly funded structures are characterized by severe rent burdens and higher poverty. The benefits of stadium development appear to skew heavily toward real estate developers and out-of-town tourists, potentially driving the displacement of low-income, long-term residents who can no longer afford the rising cost of living.
While our findings are statistically significant, several limitations within our study must be acknowledged. First, our reliance on cross-sectional, observational data provides a snapshot of the current economic climate but cannot definitively prove causation; it is entirely possible that cities selectively build stadiums in central business districts already prone to these specific economic divides. Second, the ACS dataset relies on self-reported estimates with inherent margins of error, which can be particularly high when analyzing localized ZIP codes. Additionally, our 2021 stadium location dataset may not account for recent relocations. Finally, establishing a rigid two-mile buffer assumes economic impact radiates symmetrically, ignoring how urban geography like highways, rivers, and zoning laws can isolate a nearby neighborhood from the stadium's economic ecosystem.

To build upon this research and address these limitations, future investigations should use a time series analysis. Tracking specific neighborhoods ten years before and ten years after a stadium is constructed would allow researchers to observe the exact rate of gentrification and track how quickly property values spike. Another addition to this research is filtering ACS data by industry in order to reveal whether stadium neighborhoods are oversaturated with seasonal, part-time hospitality work, which would further explain the high unemployment rates.
