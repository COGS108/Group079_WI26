## Abstract

This project examines whether neighborhoods within a 2-mile radius of professional sports stadiums in the United States have a different economic makeup than the broader cities around them. We used stadium location data, along with Census Bureau estimates on economic and housing characteristics, to categorize ZIP codes as either stadium-adjacent treatment areas or greater-city control areas. Our group compared median household income, unemployment rates, poverty rates, median home values, gross rent, vacancy rates, and new construction activity using summary statistics, data visualization, and t-tests.


Our results are somewhat mixed: stadium-adjacent neighborhoods seem to have slightly lower median household incomes, higher unemployment, and higher poverty rates than control areas, though the income differences were not statistically significant based on our t-tests. However, property values showed that home values and rents are significantly higher near stadiums, and these areas also have higher vacancy rates and more new construction, which may indicate greater real estate investment and gentrification in these areas. Overall, stadium neighborhoods appear to have distinct and variable economies, with simultaneously higher real estate costs and lower residential wealth.


These findings have real implications for housing policies and legislation, specifically regarding affordability and neighborhood equity. When cities approve stadium development, surrounding communities may bear costs like increased rents, increased vacancy, and gradual displacement of residents. Policymakers should treat stadium development not just as an economic opportunity, but as a potential community investment with participation and input from existing residents to maximize benefit while minimizing harm.

## Background and Prior Work

While stadiums may generate a localized economic surge within their immediate vicinity<sup>[1](#https://www.aimspress.com/article/doi/10.3934/NAR.2024024?viewType=HTML)</sup> and foster civic identity and pride, analysts argue their regional economic impact is limited.

### Mapping Stadiums to Cities and ZIP Codes
Connects raw stadium coordinates to specific geographic identifiers: Zip Codes and City Names

- Spatial Alignment: Convert latitude and longitude coordinates into point objects and perform a spatial join with Census ZCTA boundaries to identify which ZIP code each stadium is in

- Data Enrichment: Merge the spatial results with data from uszips to attach official city and state names

```python
#load uszips
df_uszips = pd.read_csv('data/00-raw/uszips.csv', compression='zip')

df_uszips['zip'] = df_uszips['zip'].astype(str).str.zfill(5)

df_uszips = df_uszips[['zip', 'city', 'state_id']]
```

```python
import geopandas as gpd
import pandas as pd

# vectorized geometry creation
gdf_stadiums = gpd.GeoDataFrame(
    df_clean_stadiums, 
    geometry=gpd.points_from_xy(df_clean_stadiums.Long, df_clean_stadiums.Lat),
    crs="EPSG:4326" 
)

# Load the shapes
gdf_zcta_shapes = gpd.read_file('zip://data/00-raw/tl_2025_us_zcta520.zip')

# Project the points
gdf_stadiums = gdf_stadiums.to_crs(gdf_zcta_shapes.crs)

# Spatial Join
stadiums_with_zips = gpd.sjoin(gdf_stadiums, gdf_zcta_shapes, how="left", predicate="within")

# Format ZIPs for merging
stadiums_with_zips['ZCTA5CE20'] = stadiums_with_zips['ZCTA5CE20'].astype(str)
df_uszips['zip'] = df_uszips['zip'].astype(str).str.zfill(5)

# Left Join to Kaggle reference
final_stadiums = stadiums_with_zips.merge(
    df_uszips[['zip', 'city', 'state_id']], 
    left_on='ZCTA5CE20', 
    right_on='zip', 
    how='left'
)

# Clean up and Rename
final_stadiums = final_stadiums.drop(columns=['zip'])
final_stadiums = final_stadiums.rename(columns={
    'ZCTA5CE20': 'Zip_Code', 
    'city': 'City', 
    'state_id': 'State'
})

# Apply critical patches
final_stadiums.loc[final_stadiums['Team'] == 'Vegas Golden Knights', ['City', 'State']] = ['Las Vegas', 'NV']
final_stadiums.loc[final_stadiums['Team'] == 'FC Dallas', ['City', 'State']] = ['Frisco', 'TX']

final_stadiums['City_State_Unique'] = final_stadiums['City'] + ", " + final_stadiums['State']

# Drop unnecessary columns
final_stadiums = final_stadiums.drop(columns=[
    'index_right', 'GEOID20', 'GEOIDFQ20', 'CLASSFP20', 
    'MTFCC20', 'FUNCSTAT20', 'INTPTLAT20', 'INTPTLON20'
])

print(f"Number of Stadiums: {len(final_stadiums)}")

missing_zips = final_stadiums['Zip_Code'].isna().sum()
missing_cities = final_stadiums['City'].isna().sum()
print(f"\nMissing ZIP Codes: {missing_zips}")
print(f"Missing Cities: {missing_cities}")

display(final_stadiums.head(10))
```

### Defining the Treatment Group:
Isolate the specific neighborhoods that are most likely to be affected by the presence of a professional sports stadium

- Spatial Buffer Analysis: Create a standardized 2 mile zone around each stadium coordinate to define the boundary for potential localized economic effects

- Neighborhood Identification: Determine which ZIP codes fall within these impact zones

- Final Treatment Group: Create list of Treatment ZIP codes

```python
import geopandas as gpd

# Project to meters and create 2 mile buffer
treatment_zones = final_stadiums.to_crs(epsg=5070)
treatment_zones['geometry'] = treatment_zones.geometry.buffer(3218.68)

# Project zones back to match the Census ZCTA coordinate system
treatment_zones_aligned = treatment_zones.to_crs(gdf_zcta_shapes.crs)

# Clean columns
cols_to_drop = ['index_right', 'index_left']
treatment_zones_aligned = treatment_zones_aligned.drop(columns=[c for c in cols_to_drop if c in treatment_zones_aligned.columns])

# Find all Census ZCTAs that touch the 2 mile stadium boundaries
stadium_neighborhoods = gpd.sjoin(
    gdf_zcta_shapes, 
    treatment_zones_aligned, 
    how="inner", 
    predicate="intersects"
)

# Extract unique ZIPs
unique_treatment_zips = stadium_neighborhoods['ZCTA5CE20'].unique()
treatment_group = pd.DataFrame({'Treatment_Zip': unique_treatment_zips})

# Ensure all ZIPs are 5 char strings
treatment_group['Treatment_Zip'] = treatment_group['Treatment_Zip'].astype(str).str.zfill(5)

print(f"Treatment Group defined with {len(treatment_group)} unique ZIP codes.")
```

### Defining the Control Group
Identify a baseline for comparison by finding neighborhoods within the same urban economies that are far enough away to be unaffected by stadium externalities.

- Identify Stadium Cities and Associated Zips: Create a master list of the 71 unique cities (and all the Zips they contain) that host the professional stadiums in the study

- Exclude the Treatment Group: Remove any neighborhoods that fall within the 2 mile stadium radius

- Final Control Group: Create List

```python
# 1. Identify all ZIPs belonging to your 71 stadium cities
stadium_cities = final_stadiums['City_State_Unique'].unique()
df_uszips['City_State_Unique'] = df_uszips['city'] + ", " + df_uszips['state_id']
city_zips = df_uszips[df_uszips['City_State_Unique'].isin(stadium_cities)].copy()
city_zips['zip'] = city_zips['zip'].astype(str).str.zfill(5)

# 2. Extract the treatment list for filtering
treatment_zip_list = treatment_group['Treatment_Zip'].unique()

# 3. Create control group by excluding treatment ZIPs from the city list
control_group = city_zips[~city_zips['zip'].isin(treatment_zip_list)].copy()

# 4. Clean and Rename
control_group = control_group.rename(columns={'zip': 'Control_Zip'})

print("Control Group Summary:")
print(f"Total Stadium Cities: {len(stadium_cities)}")
print(f"Total ZIPs in these cities: {len(city_zips)}")
print(f"Treatment ZIPs within these cities: {len(city_zips[city_zips['zip'].isin(treatment_zip_list)])}")
print(f"Control ZIPs: {len(control_group)}")
```

```python
# Label the groups for the final analysis
treatment_group['Group'] = 'Treatment (2-Mile Radius)'
control_group['Group'] = 'Control (Rest of City)'

# Standardize column names for stacking
t_list = treatment_group.rename(columns={'Treatment_Zip': 'zip_code'})[['zip_code', 'Group']]
c_list = control_group.rename(columns={'Control_Zip': 'zip_code'})[['zip_code', 'Group']]

# Stack them into one master lookup table
combined_groups = pd.concat([t_list, c_list], ignore_index=True)

# Final safety check: ensure no duplicates if a ZIP was somehow in both lists
combined_groups = combined_groups.drop_duplicates(subset=['zip_code'])

print(f"Master Key 'combined_groups' created with {len(combined_groups)} total ZIPs.")
print(combined_groups['Group'].value_counts())
```

### Data Set #3: ACS 2024 5-Year Estimates Selected Housing Characteristics (DP04)

This dataset is from the U.S. Census Bureau’s American Community Survey (ACS) 2024 5-Year Estimates. It provides a comprehensive profile of the housing stock, financial costs of homeownership and renting, and physical characteristics of housing units across specific geographies. The ACS is an ongoing survey that provides characteristics data, helping planners, researchers, and real estate analysts understand the quality of life and economic burden of housing in specific regions.

The dataset includes several critical metrics for housing market analysis. Median Home Value and Median Gross Rent are provided in US Dollars, they represent the midpoint of all reported values, where half are higher and half are lower. A healthy housing market typically sees these values rising in line with local inflation, whereas stagnant or falling values may indicate economic decline. Housing Tenure (Owner vs. Renter occupied) is measured in raw counts and percentages, a high owner-occupancy rate suggests neighborhood stability, while high renter-occupancy can indicate a transient population or an area with high density multi-family development. Another important metric is Gross Rent as a Percentage of Household Income (GRAPI). The Census Bureau generally considers households spending 30% or more of their income on housing to be "rent burdened," while those spending 50% or more are "severely rent burdened," indicating a potential crisis in affordability.

A major concern with this dataset is its reliance on self-reporting and sampling error. Because this is a survey of a sample of the population rather than a full count, every estimate comes with a Margin of Error (MOE). In smaller geographic areas (like specific census tracts near a stadium), the MOE can be quite large relative to the estimate, making the data less reliable. There is also a risk of Non-response Bias, when certain demographics, such as very high-income individuals or undocumented residents, may be less likely to respond to the survey accurately. Finally, "Median Value" is based on the respondent’s perception of what their home is worth, which may differ from actual market sale prices or tax appraisals.

```python
# delete imports if done already
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

def _first_series(df, colname):
    selected = df.loc[:, colname]
    if isinstance(selected, pd.DataFrame):
        return selected.iloc[:, 0]
    return selected

def _to_num(series):
    return pd.to_numeric(
        series.astype(str)
        .str.replace(',', '', regex=False)
        .str.replace('$', '', regex=False)
        .str.replace('(X)', '', regex=False)
        .str.replace('nan', '', regex=False),
        errors='coerce'
    )

# Try direct median household income column -----
income_candidates = [
    c for c in df_income.columns
    if ('Median household income (dollars)' in c) and ('Margin' not in c)
]

direct_income_df = pd.DataFrame()
if income_candidates:
    chosen_col = income_candidates[0]
    direct_income_df = pd.DataFrame({
        'zip_code': _first_series(df_income, 'Geography').astype(str).str.extract(r'(\d{5})', expand=False).str.zfill(5),
        'median_household_income': _to_num(_first_series(df_income, chosen_col))
    }).dropna(subset=['zip_code', 'median_household_income'])

# Fallback: estimate median from income-bin percentages 
if direct_income_df.empty:
    bin_specs = [
        ('Less than $10,000', 0, 10000),
        ('$10,000 to $14,999', 10000, 15000),
        ('$15,000 to $24,999', 15000, 25000),
        ('$25,000 to $34,999', 25000, 35000),
        ('$35,000 to $49,999', 35000, 50000),
        ('$50,000 to $74,999', 50000, 75000),
        ('$75,000 to $99,999', 75000, 100000),
        ('$100,000 to $149,999', 100000, 150000),
        ('$150,000 to $199,999', 150000, 200000),
        ('$200,000 or more', 200000, 250000),
    ]

    bin_cols = []
    for label, _, _ in bin_specs:
        matches = [
            c for c in df_income.columns
            if ('Total households!!' + label in c) and ('Margin' not in c)
        ]
        if matches:
            bin_cols.append(matches[0])
        else:
            bin_cols.append(None)

    if any(c is None for c in bin_cols):
        missing_labels = [bin_specs[i][0] for i, c in enumerate(bin_cols) if c is None]
        raise ValueError(f'Missing required income-bin columns for fallback estimate: {missing_labels}')

    pct_matrix = pd.concat([_to_num(_first_series(df_income, c)).rename(c) for c in bin_cols], axis=1).fillna(0.0)

    def _estimate_median_from_bins(row_vals):
        percentages = np.array(row_vals, dtype=float)
        total = np.nansum(percentages)
        if total <= 0:
            return np.nan
        percentages = percentages * (100.0 / total)
        cumulative = 0.0
        for pct, (_, lower, upper) in zip(percentages, bin_specs):
            next_cum = cumulative + pct
            if next_cum >= 50:
                if pct <= 0:
                    return float(lower)
                frac = (50 - cumulative) / pct
                return float(lower + frac * (upper - lower))
            cumulative = next_cum
        return float(bin_specs[-1][2])

    estimated_income = pct_matrix.apply(_estimate_median_from_bins, axis=1)

    direct_income_df = pd.DataFrame({
        'zip_code': _first_series(df_income, 'Geography').astype(str).str.extract(r'(\d{5})', expand=False).str.zfill(5),
        'median_household_income': estimated_income
    }).dropna(subset=['zip_code', 'median_household_income'])

    # Use fallback estimate from household income bins because direct median income column had no numeric values.

#  Merge with treatment/control groups and plot
groups_for_plot = combined_groups.copy()
groups_for_plot['zip_code'] = groups_for_plot['zip_code'].astype(str).str.zfill(5)

income_violin_df = direct_income_df.merge(
    groups_for_plot[['zip_code', 'Group']],
    on='zip_code',
    how='inner'
 )


if income_violin_df.empty:
    print('No overlapping rows found for violin plot.')
else:
    plt.figure(figsize=(10, 6))
    sns.violinplot(
        data=income_violin_df,
        x='Group',
        y='median_household_income',
        inner='quartile',
        cut=0
    )
    sns.stripplot(
        data=income_violin_df.sample(min(800, len(income_violin_df)), random_state=42),
        x='Group',
        y='median_household_income',
        color='black',
        alpha=0.25,
        size=2
    )
    plt.title('Median Household Income Distribution by Group')
    plt.xlabel('Group')
    plt.ylabel('Median Household Income (USD)')
    plt.xticks(rotation=8)
    plt.tight_layout()
    plt.show()

    print('\nSummary stats by group:')
    display(
        income_violin_df.groupby('Group')['median_household_income']
        .describe()[['count', 'mean', '50%', 'std', 'min', 'max']]
        .rename(columns={'50%': 'median'})
        .round(2)
    )
```

## Discusison and Conclusion

Our research found that neighborhoods within two miles of U.S. sports stadiums fundamentally contradict our initial hypotheses of job growth and lower property values. Instead, stadium proximity is linked to a roughly $105,700 premium on median home values (p<0.00001), significantly higher rents (p=0.00005), and increased new development and vacancy rates (p<0.05). Most notably, rather than boosting the local economy for residents, these stadium-adjacent neighborhoods face significantly higher average unemployment (6.79% vs. 5.84%, p=0.00002) and higher mean poverty rates (18.02% vs. 14.27%), while median household incomes remain statistically unchanged compared to the rest of the city (p=0.439).


Ultimately, these metrics point to the results of gentrification and speculative investment rather than broad community uplift. Stadium neighborhoods are more expensive to live in, attracting new development and short-term or vacant investor properties. However, this localized influx of wealth does not translate into economic prosperity for residents. Flat incomes combined with expensive housing and a 10% vacancy rate show a clear disconnect: the neighborhood is getting richer, but the residents are not. The people purchasing these expensive, newly developed properties likely do not live there full-time, leaving the long-term, local residents to navigate higher unemployment and poverty rates than the citywide average.


The broader skepticism found in our background literature aligned closely with these findings. Bradbury, Coates, and Humphreys (2024) argued that the economic improvements promised by new stadiums are often vastly overstated compared to their massive public costs, a sentiment our unemployment data strongly supports. Furthermore, our property valuation findings add a modern layer to Charles Tu’s (2005) observations on stadium-adjacent real estate. While Tu noted that homes might initially sell for less before land values eventually rise, our 2024 ACS data suggests we are currently observing the end of that timeline. Stadiums appear to act as anchors for luxury urban development and out-of-town investment, echoing Valerio's (2024) insights on stadiums driving localized economic production in a way that prioritizes real estate capital over resident welfare.


This highlights an ethical issue regarding urban planning and public finance. Professional sports stadiums are frequently subsidized by taxpayer dollars under the promise of job creation and community revitalization. However, our data suggests a failure to deliver on this promise for the immediate community. The areas directly surrounding these publicly funded structures are characterized by severe rent burdens and higher poverty. The benefits of stadium development appear to skew heavily toward real estate developers and out-of-town tourists, potentially driving the displacement of low-income, long-term residents who can no longer afford the rising cost of living.

While our findings are statistically significant, several limitations within our study must be acknowledged. First, our reliance on cross-sectional, observational data provides a snapshot of the current economic climate but cannot definitively prove causation; it is entirely possible that cities selectively build stadiums in central business districts already prone to these specific economic divides. Second, the ACS dataset relies on self-reported estimates with inherent margins of error, which can be particularly high when analyzing localized ZIP codes. Additionally, our 2021 stadium location dataset may not account for recent relocations. Finally, establishing a rigid two-mile buffer assumes economic impact radiates symmetrically, ignoring how urban geography like highways, rivers, and zoning laws can isolate a nearby neighborhood from the stadium's economic ecosystem.


To build upon this research and address these limitations, future investigations should use a time series analysis. Tracking specific neighborhoods ten years before and ten years after a stadium is constructed would allow researchers to observe the exact rate of gentrification and track how quickly property values spike. Another addition to this research is filtering ACS data by industry in order to reveal whether stadium neighborhoods are oversaturated with seasonal, part-time hospitality work, which would further explain the high unemployment rates.
