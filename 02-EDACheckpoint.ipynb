### Dataset #1: Sports Stadium Locations

1. Important metrics: Latitude and longitude coordinates denote the horizontal and vertical coordinates of the stadium on Earth. They are composed of decimal degrees and, when put together, can pinpoint the exact location of a given sports stadium. These can be used to provide specific location information regarding the stadium that Python can parse through. This also helps clarify inaccurate team names that reflect a nearby major city or the name of the state rather than the physical location of the stadium.

2. Major concerns: This dataset is from 2021 and may not reflect accurate team locations due to the movement of sports teams within the last 5 years. Additionally, its exhaustiveness (i.e., whether or not it includes all professional sports stadiums) has not been confirmed.

```python
import pandas as pd
import numpy as np
import os

# Setup
raw_data_path = 'data/00-raw/stadiums.csv'
processed_data_path = 'data/02-processed/processed_data.csv'

os.makedirs('data/02-processed', exist_ok=True)

try:
    df_raw_stadiums = pd.read_csv(raw_data_path, low_memory=False)
except FileNotFoundError:
    print('Files not found. Please ensure they are in data/00-raw/')
    df_raw_stadiums = pd.DataFrame()

# Tidy Structure
print('First 5 Rows:')
print(df_raw_stadiums.head())
print('Column Names:')
print(df_raw_stadiums.columns.tolist())
print('Each row represents one professional sports team and each column represents a variable, such as league and coordinates.')
print('Dataset is already tidy.')

# Shape
print('Dataset Shape (rows, columns):', df_raw_stadiums.shape)

# Missing Values
missing_vals = df_raw_stadiums.isna().sum()
print('Missing Values Per Column:')
print(missing_vals)
print('Missingness Correlation Matrix:')
print(df_raw_stadiums.isna().astype(int).corr())

# Location Outliers
numeric_cols = ['Lat', 'Long']
for col in numeric_cols:
    Q1 = df_raw_stadiums[col].quantile(0.25)
    Q3 = df_raw_stadiums[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df_raw_stadiums[(df_raw_stadiums[col] < lower) | (df_raw_stadiums[col] > upper)]
    print(f'Outliers in {col}: {len(outliers)}')
# outliers in {col}: {len(outliers)}')

# Data Cleaning
df_clean_stadiums = df_raw_stadiums.dropna(subset=['Lat', 'Long'])

# Remove Canadian teams (falling outside US Census ZCTA boundaries)
canadian_teams = [
    'Toronto Raptors', 'Montreal Canadiens', 'Winnipeg Jets', 
    'Ottawa Senators', 'Vancouver Canucks', 'Edmonton Oilers', 
    'Toronto Maple Leafs', 'Calgary Flames', 'Toronto Blue Jays', 
    'Vancouver Whitecaps FC', 'Toronto FC', 'CF Montréal'
]
df_clean_stadiums = df_clean_stadiums[~df_clean_stadiums['Team'].isin(canadian_teams)]

print('Final Dataset Shape After Cleaning:', df_clean_stadiums.shape)

# Save Processed Data
df_clean_stadiums.to_csv(processed_data_path, index=False)
print('Processed dataset saved to data/02-processed/')
```

### Mapping Stadiums to Cities and ZIP Codes
Connects raw stadium coordinates to specific geographic identifiers: Zip Codes and City Names

- Spatial Alignment: Convert latitude and longitude coordinates into point objects and perform a spatial join with Census ZCTA boundaries to identify which ZIP code each stadium is in

- Data Enrichment: Merge the spatial results with data from uszips to attach official city and state names

```python
#load uszips
df_uszips = pd.read_csv('data/00-raw/uszips.csv', compression='zip')

df_uszips['zip'] = df_uszips['zip'].astype(str).str.zfill(5)

df_uszips = df_uszips[['zip', 'city', 'state_id']]
```

```python
import geopandas as gpd
import pandas as pd

# vectorized geometry creation
gdf_stadiums = gpd.GeoDataFrame(
    df_clean_stadiums, 
    geometry=gpd.points_from_xy(df_clean_stadiums.Long, df_clean_stadiums.Lat),
    crs="EPSG:4326" 
)

# Load the shapes
gdf_zcta_shapes = gpd.read_file('zip://data/00-raw/tl_2025_us_zcta520.zip')

# Project the points
gdf_stadiums = gdf_stadiums.to_crs(gdf_zcta_shapes.crs)

# Spatial Join
stadiums_with_zips = gpd.sjoin(gdf_stadiums, gdf_zcta_shapes, how="left", predicate="within")

# Format ZIPs for merging
stadiums_with_zips['ZCTA5CE20'] = stadiums_with_zips['ZCTA5CE20'].astype(str)
df_uszips['zip'] = df_uszips['zip'].astype(str).str.zfill(5)

# Left Join to Kaggle reference
final_stadiums = stadiums_with_zips.merge(
    df_uszips[['zip', 'city', 'state_id']], 
    left_on='ZCTA5CE20', 
    right_on='zip', 
    how='left'
)

# Clean up and Rename
final_stadiums = final_stadiums.drop(columns=['zip'])
final_stadiums = final_stadiums.rename(columns={
    'ZCTA5CE20': 'Zip_Code', 
    'city': 'City', 
    'state_id': 'State'
})

# Apply critical patches
final_stadiums.loc[final_stadiums['Team'] == 'Vegas Golden Knights', ['City', 'State']] = ['Las Vegas', 'NV']
final_stadiums.loc[final_stadiums['Team'] == 'FC Dallas', ['City', 'State']] = ['Frisco', 'TX']

final_stadiums['City_State_Unique'] = final_stadiums['City'] + ", " + final_stadiums['State']

# Drop unnecessary columns
final_stadiums = final_stadiums.drop(columns=[
    'index_right', 'GEOID20', 'GEOIDFQ20', 'CLASSFP20', 
    'MTFCC20', 'FUNCSTAT20', 'INTPTLAT20', 'INTPTLON20'
])

print(f"Number of Stadiums: {len(final_stadiums)}")

missing_zips = final_stadiums['Zip_Code'].isna().sum()
missing_cities = final_stadiums['City'].isna().sum()
print(f"\nMissing ZIP Codes: {missing_zips}")
print(f"Missing Cities: {missing_cities}")

display(final_stadiums.head(10))
```

### Defining the Treatment Group:
Isolate the specific neighborhoods that are most likely to be affected by the presence of a professional sports stadium

- Spatial Buffer Analysis: Create a standardized 2 mile zone around each stadium coordinate to define the boundary for potential localized economic effects

- Neighborhood Identification: Determine which ZIP codes fall within these impact zones

- Final Treatment Group: Create list of Treatment ZIP codes

```python
import geopandas as gpd

# Project to meters and create 2 mile buffer
treatment_zones = final_stadiums.to_crs(epsg=5070)
treatment_zones['geometry'] = treatment_zones.geometry.buffer(3218.68)

# Project zones back to match the Census ZCTA coordinate system
treatment_zones_aligned = treatment_zones.to_crs(gdf_zcta_shapes.crs)

# Clean columns
cols_to_drop = ['index_right', 'index_left']
treatment_zones_aligned = treatment_zones_aligned.drop(columns=[c for c in cols_to_drop if c in treatment_zones_aligned.columns])

# Find all Census ZCTAs that touch the 2 mile stadium boundaries
stadium_neighborhoods = gpd.sjoin(
    gdf_zcta_shapes, 
    treatment_zones_aligned, 
    how="inner", 
    predicate="intersects"
)

# Extract unique ZIPs
unique_treatment_zips = stadium_neighborhoods['ZCTA5CE20'].unique()
treatment_group = pd.DataFrame({'Treatment_Zip': unique_treatment_zips})

# Ensure all ZIPs are 5 char strings
treatment_group['Treatment_Zip'] = treatment_group['Treatment_Zip'].astype(str).str.zfill(5)

print(f"Treatment Group defined with {len(treatment_group)} unique ZIP codes.")
```

```python
# 1. Identify all ZIPs belonging to your 71 stadium cities
stadium_cities = final_stadiums['City_State_Unique'].unique()
df_uszips['City_State_Unique'] = df_uszips['city'] + ", " + df_uszips['state_id']
city_zips = df_uszips[df_uszips['City_State_Unique'].isin(stadium_cities)].copy()
city_zips['zip'] = city_zips['zip'].astype(str).str.zfill(5)

# 2. Extract the treatment list for filtering
treatment_zip_list = treatment_group['Treatment_Zip'].unique()

# 3. Create control group by excluding treatment ZIPs from the city list
control_group = city_zips[~city_zips['zip'].isin(treatment_zip_list)].copy()

# 4. Clean and Rename
control_group = control_group.rename(columns={'zip': 'Control_Zip'})

print("Control Group Summary:")
print(f"Total Stadium Cities: {len(stadium_cities)}")
print(f"Total ZIPs in these cities: {len(city_zips)}")
print(f"Treatment ZIPs within these cities: {len(city_zips[city_zips['zip'].isin(treatment_zip_list)])}")
print(f"Control ZIPs: {len(control_group)}")
```

```python
# Label the groups for the final analysis
treatment_group['Group'] = 'Treatment (2-Mile Radius)'
control_group['Group'] = 'Control (Rest of City)'

# Standardize column names for stacking
t_list = treatment_group.rename(columns={'Treatment_Zip': 'zip_code'})[['zip_code', 'Group']]
c_list = control_group.rename(columns={'Control_Zip': 'zip_code'})[['zip_code', 'Group']]

# Stack them into one master lookup table
combined_groups = pd.concat([t_list, c_list], ignore_index=True)

# Final safety check: ensure no duplicates if a ZIP was somehow in both lists
combined_groups = combined_groups.drop_duplicates(subset=['zip_code'])

print(f"Master Key 'combined_groups' created with {len(combined_groups)} total ZIPs.")
print(combined_groups['Group'].value_counts())
```
